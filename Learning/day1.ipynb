```python
# Lets Goooo

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        return f"Hello, {self.name}! You are {self.age} years old."


p = Person("Krishna Mandal", 24)
p.greet()

print("Hello")
```
